```python
import pandas as pd
import numpy as np

# 1. Load Data Test (Kunci Jawaban Masa Depan)
test_path = '../data/processed/test.csv'
df_test = pd.read_csv(test_path)

# 2. Filter hanya interaksi positif (Misal Rating >= 4) sebagai barang yang "Relevan"
df_test_relevant = df_test[df_test['rating'] >= 4]

# 3. Buat Kunci Jawaban: Mengelompokkan ID Item yang disukai berdasarkan User ID
# Format output: {user_id_1: {item_A, item_B}, user_id_2: {item_C}, ...}
actual_interactions = df_test_relevant.groupby('user_id')['item_id'].apply(set).to_dict()

print(f"Total user di data test yang memiliki interaksi positif: {len(actual_interactions)}")
```

```python
def calculate_precision_recall_at_k(recommended_items, actual_items, k=10):
    """
    Menghitung Precision@K dan Recall@K
    """
    # Ambil hanya Top K dari rekomendasi
    top_k_recs = recommended_items[:k]
    
    # Hitung jumlah "Hits" (Berapa barang rekomendasi yang cocok dengan kunci jawaban)
    hits = len(set(top_k_recs).intersection(actual_items))
    
    # Precision: Dari K tebakan, berapa yang benar?
    precision = hits / k if k > 0 else 0.0
    
    # Recall: Dari total barang relevan, berapa yang berhasil ditebak?
    recall = hits / len(actual_items) if len(actual_items) > 0 else 0.0
    
    return precision, recall

print("Fungsi metrik evaluasi siap digunakan!")
```

```python
# 1. Kita ambil satu User ID secara acak TAPI yang PASTI ADA di dalam kunci jawaban
user_test_id = list(actual_interactions.keys())[5] # Mengambil user ke-5 dari data test

print(f"=== MENGUJI USER {user_test_id} ===")

# 2. Asumsikan ini tebakan Top 10 dari model kita (Kita campur tebakan acak dengan jawaban asli agar skornya tidak 0)
jawaban_asli_user_ini = list(actual_interactions[user_test_id])
simulated_top_10 = jawaban_asli_user_ini[:3] + [999, 888, 777, 666, 555, 444, 333] # 3 tebakan benar, 7 salah

actual_items = actual_interactions[user_test_id]
print(f"Kunci Jawaban (Masa Depan): {list(actual_items)[:10]}...")
print(f"Tebakan Sistem (Top 10): {simulated_top_10}")
    
# 3. Lakukan evaluasi
precision, recall = calculate_precision_recall_at_k(simulated_top_10, actual_items, k=10)
    
print("\n=== HASIL EVALUASI TOP-10 ===")
print(f"Precision@10 : {precision:.2f}")
print(f"Recall@10    : {recall:.2f}")
```
