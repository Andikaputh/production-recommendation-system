```python
import pandas as pd
import os

# 1. Memuat Data Mentah
ratings_path = '../data/raw/ml-100k/u.data'
df = pd.read_csv(ratings_path, sep='\t', names=['user_id', 'item_id', 'rating', 'timestamp'])

# 2. Membersihkan Data (Transformasi Waktu)
# Mengubah timestamp angka menjadi format waktu datetime yang sebenarnya
df['timestamp'] = pd.to_datetime(df['timestamp'], unit='s')

# 3. Mengurutkan seluruh data berdasarkan waktu terjadinya interaksi (Kronologis)
df = df.sort_values('timestamp').reset_index(drop=True)

display(df.head())
print(f"Transaksi pertama: {df['timestamp'].min()}")
print(f"Transaksi terakhir: {df['timestamp'].max()}")
```

```python
# Menentukan titik potong 80%
split_index = int(len(df) * 0.8)

# Membagi data
df_train = df.iloc[:split_index].copy()
df_test = df.iloc[split_index:].copy()

print(f"Jumlah Data Train (80% terdahulu): {len(df_train)}")
print(f"Jumlah Data Test (20% terbaru): {len(df_test)}")

# Validasi bahwa tidak ada kebocoran waktu
print("\nValidasi Pemisahan Waktu:")
print(f"Waktu terakhir di Train: {df_train['timestamp'].max()}")
print(f"Waktu pertama di Test: {df_test['timestamp'].min()}")
```

```python
# Membuat folder processed jika belum ada
PROCESSED_DIR = '../data/processed'
os.makedirs(PROCESSED_DIR, exist_ok=True)

# Menyimpan hasil pipeline
train_path = os.path.join(PROCESSED_DIR, 'train.csv')
test_path = os.path.join(PROCESSED_DIR, 'test.csv')

df_train.to_csv(train_path, index=False)
df_test.to_csv(test_path, index=False)

print(f"Data Pipeline Berhasil! Data tersimpan di:\n1. {train_path}\n2. {test_path}")
```
